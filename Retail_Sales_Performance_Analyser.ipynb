## 1. Problem Definition

A retailer requires a Python program to analyse sales transactions and produce transaction, branch, product and salesperson performance summaries.

The supplied retail sales dataset contains information about sale IDs, branches, products, product categories, quantities, unit prices, payment methods and salespersons. Before performing any analysis, each transaction must be validated according to the rules specified in the capstone exercise.

The program must identify and report invalid transactions, including the reason for rejection. Recognized capitalization differences in branch names must be standardized before the transactions are analysed.

For valid transactions, the program will calculate transaction values and produce sales performance summaries. The program will also provide a menu through which the user can view transactions, analyse sales, compare branches, view salesperson performance, view invalid records and view the overall summary.

The solution will be implemented using Core Python programming concepts including variables, decisions, loops, data structures, functions, validation and menu control.

## 2. Objectives

The objectives of the Retail Sales Performance Analyser are:

1. To validate every retail sales transaction according to the specified validation rules.

2. To standardize recognized capitalization differences in branch names.

3. To identify and report invalid transactions together with their rejection reasons.

4. To calculate the transaction value for each valid transaction using quantity multiplied by unit price.

5. To calculate total and average transaction value and identify the highest-value transaction.

6. To identify the branch, product and salesperson generating the highest revenue.

7. To count transactions by payment method.

8. To compare branch performance.

9. To display transactions exceeding a user-provided threshold.

10. To provide a menu-driven Python program that allows users to access the required operations.

## 3. Standardise the branch names

S017 has "westlands" in lower case, which would be counted as a different branch
from "Westlands". Changing every branch name to upper case fixes it.

Upper case is used rather than title case, because `.title()` would turn
"Nairobi CBD" into "Nairobi Cbd".

```python
#Change every branch name to upper case
def standardise_branches(sales):
  for i in range(len(sales)):
    sales[i]['branch'] = sales[i]['branch'].upper()
  return sales


dict_sales = standardise_branches(dict_sales)

# Show that S017 now matches the other Westlands records.
print(dict_sales[0]['branch'])
print(dict_sales[16]['branch'])
```

## 4. Validate the records

The four rules from the brief:

1. Sale ID, branch and product are provided
2. Quantity is numeric and greater than zero
3. Unit price is numeric and greater than zero
4. Payment method is recognised

Records that pass go into `valid_sales` and get an `amount` added, worked out as
quantity times unit price. Records that fail go into `invalid_sales` with a
reason, so option 5 can report them.

```python
#Split the records into valid ones and invalid ones
def validate_sales(sales):
  valid_sales = []
  invalid_sales = []

  for i in range(len(sales)):
    record = sales[i]
    reason = ""

    # Rule 1: sale id, branch and product must be provided.
    if record['sale_id'] == "":
      reason = "sale id is missing"
    elif record['branch'] == "":
      reason = "branch is missing"
    elif record['product'] == "":
      reason = "product is missing"

    # Rule 2: quantity numeric and greater than zero.
    elif type(record['quantity']) != int and type(record['quantity']) != float:
      reason = "quantity is not a number"
    elif record['quantity'] <= 0:
      reason = "quantity is not greater than zero"

    # Rule 3: unit price numeric and greater than zero.
    elif type(record['unit_price_kes']) != int and type(record['unit_price_kes']) != float:
      reason = "unit price is not a number"
    elif record['unit_price_kes'] <= 0:
      reason = "unit price is not greater than zero"

    # Rule 4: payment method must be one the retailer accepts.
    elif record['payment_method'] not in payment_methods:
      reason = "payment method is not recognised"

    if reason == "":
      record['amount'] = record['quantity'] * record['unit_price_kes']
      valid_sales.append(record)
    else:
      record['reason'] = reason
      invalid_sales.append(record)

  return valid_sales, invalid_sales


valid_sales, invalid_sales = validate_sales(dict_sales)

print(f"Records read: {len(dict_sales)}")
print(f"Valid records: {len(valid_sales)}")
print(f"Invalid records: {len(invalid_sales)}")
```

## 5. The calculations

Six small functions, each doing one job. They are kept separate from the
printing so they can be tested on their own in section 7.

```python
#Add up the amount of every record
def total_sales(sales):
  total = 0
  for i in range(len(sales)):
    total = total + sales[i]['amount']
  return total

#Work out the average amount. Returns 0 if the list is empty
def average_sales(sales):
  if len(sales) == 0:
    return 0
  return total_sales(sales) / len(sales)

#Find the record with the biggest amount
def highest_sale(sales):
  if len(sales) == 0:
    return None
  highest = sales[0]
  for i in range(len(sales)):
    if sales[i]['amount'] > highest['amount']:
      highest = sales[i]
  return highest


print(f"Total: {total_sales(valid_sales)}")
print(f"Average: {average_sales(valid_sales): .2f}")
print(f"Highest: {highest_sale(valid_sales)['sale_id']}")
```

```python
#Add up the revenue for each branch, product or salesperson.
#The field is passed in, so the same function works for all three instead of writing it out three times.
def revenue_totals(sales, field):
  totals = {}
  for i in range(len(sales)):
    key = sales[i][field]
    if key in totals:
      totals[key] = totals[key] + sales[i]['amount']
    else:
      totals[key] = sales[i]['amount']
  return totals

#Find the name with the biggest total
def top_earner(totals):
  top_name = ""
  top_amount = 0
  for key in totals:
    if totals[key] > top_amount:
      top_amount = totals[key]
      top_name = key
  return top_name, top_amount

#Count how many records each branch, product or salesperson has
def count_sales(sales, field):
  counts = {}
  for i in range(len(sales)):
    key = sales[i][field]
    if key in counts:
      counts[key] = counts[key] + 1
    else:
      counts[key] = 1
  return counts


print(revenue_totals(valid_sales, 'branch'))
print(top_earner(revenue_totals(valid_sales, 'branch')))
print(count_sales(valid_sales, 'branch'))
```

```python
# Calculate transaction value
def calculate_transaction_value(sales):
    # Calculate the value of one transaction
    return sales["quantity"] * sales["unit_price_kes"]
# Calculate the value of the first valid transaction
value = calculate_transaction_value(valid_sales[0])
```

```python
#Revenue by Product
def calculate_product_revenue(valid_sales):
    # Create an empty dictionary for product revenue
    product_revenue = {}

    # Process each valid transaction
    for transaction in valid_sales:
        product = transaction["product"]
        value = calculate_transaction_value(transaction)

        if product in product_revenue:
            product_revenue[product] += value
        else:
            product_revenue[product] = value

    return product_revenue
product_revenue = calculate_product_revenue(valid_sales)

print("Revenue by product:")

for product, revenue in product_revenue.items():
    print(product, ": KES", revenue)
```

```python
# Identify the Best-Selling Product
def find_best_product(valid_sales):
    product_revenue = calculate_product_revenue(valid_sales)

    if len(product_revenue) == 0:
        return None, 0

    best_product = None
    highest_revenue = 0

    for product, revenue in product_revenue.items():
        if revenue > highest_revenue:
            highest_revenue = revenue
            best_product = product

    return best_product, highest_revenue
best_product, best_product_revenue = find_best_product(valid_sales)

print("Best-Selling Product:", best_product)
print("Revenue: KES", best_product_revenue)
```

```python
# Revenue by Salesperson
def calculate_salesperson_revenue(valid_sales):
    # Create an empty dictionary for salesperson revenue
    salesperson_revenue = {}

    # Process each valid transaction
    for transaction in valid_sales:
        salesperson = transaction["salesperson"]
        value = calculate_transaction_value(transaction)

        if salesperson in salesperson_revenue:
            salesperson_revenue[salesperson] += value
        else:
            salesperson_revenue[salesperson] = value

    return salesperson_revenue
salesperson_revenue = calculate_salesperson_revenue(valid_sales)

print("Revenue by salesperson:")

for salesperson, revenue in salesperson_revenue.items():
    print(salesperson, ": KES", revenue)
```

```python
# Identify the Top Salesperson
def find_top_salesperson(valid_sales):
    salesperson_revenue = calculate_salesperson_revenue(valid_sales)

    if len(salesperson_revenue) == 0:
        return None, 0

    top_salesperson = None
    highest_revenue = 0

    for salesperson, revenue in salesperson_revenue.items():
        if revenue > highest_revenue:
            highest_revenue = revenue
            top_salesperson = salesperson

    return top_salesperson, highest_revenue
top_salesperson, top_salesperson_revenue = find_top_salesperson(valid_sales)

print("Top Salesperson:", top_salesperson)
print("Revenue: KES", top_salesperson_revenue)
```
